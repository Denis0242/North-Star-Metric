# EDA, Cleaning & Feature Engineering
North Star Metric & Product Health Dashboard

```python
import pandas as pd
from pathlib import Path

DATA_PATH = Path('../data/raw/nsm_dataset_raw.csv')
df = pd.read_csv(DATA_PATH)
df.head()
```

```python
df.shape, df.info()
```

```python
df.isna().sum().sort_values(ascending=False)
```

```python
df.duplicated().sum()
```

```python
df['event_date'] = pd.to_datetime(df['event_date'])
for col in ['country','segment','device','acquisition_channel','primary_feature_used','experiment_group','engagement_band']:
    df[col] = df[col].astype(str).str.strip().str.title()
df.head()
```

```python
rate_cols = ['stickiness_ratio','conversion_rate','d1_retention','d7_retention','d30_retention','churn_rate','activation_rate','feature_adoption_rate']
df[rate_cols].describe()
```

```python
df['retention_gap'] = df['d1_retention'] - df['d30_retention']
df['conversion_revenue_index'] = df['conversion_rate'] * df['revenue_per_user']
df['health_signal'] = df.apply(lambda r: 'Healthy' if r['d30_retention'] >= .40 and r['churn_rate'] <= .70 else 'Monitor', axis=1)
df[['retention_gap','conversion_revenue_index','health_signal']].head()
```

```python
df.to_csv('../data/processed/nsm_dataset_clean.csv', index=False)
```
